```python
import numpy as np
import matplotlib.pyplot as plt
import polars as pl
```

## Estimating Standard Deviation

$$
\sigma \approx \sqrt{\frac{\sum{ (X_i - \bar{X}_n) }}{n - 1}}
$$

```python
filename_base = "../experimental_methods/models/training_10_prey/run3/"

# this is the seed we will use for reproducibility
root_seed = 42
ss = np.random.SeedSequence(root_seed)

# Spawn 10 independent child seeds (one for each replicate of your 2^k design)
child_seeds = ss.spawn(10)

# To get the actual integer to pass to your MARL env:
seeds = [s.generate_state(1)[0] for s in child_seeds]

print(len(seeds))
print("Seeds with high entropy:", seeds)

seeds = [42, 43, 44, 45, 46, 47, 48, 100, 105, 1000]
# seeds = [127, 8191, 131071, 524287, 2147483647,
#          9582731, 422189, 73856093, 19260817, 715225741]



means = {
    'dos_means': [],
    'doa_means': [],
}

for seed in seeds:
    filename = filename_base + f"eval_metrics_{seed}.csv"
    data = pl.read_csv(filename)
    # compute mean and std of DoS and DoA
    means['dos_means'].append(data['dos'].mean())
    means['doa_means'].append(data['doa'].mean())
```

## Calculating Minimum Detectable Effect (MDE)

$$

\text{MDE} = ( z_{\alpha/2} + z_{\beta}) \frac{\sigma}{\sqrt{n}}

$$

```python
# compute standard deviation of DoS and DoA
term1 = 1 / (len(seeds) - 1)
dos_mean = np.mean(means['dos_means'])
doa_mean = np.mean(means['doa_means'])
dos_std = np.sqrt(
    term1 * np.sum((means['dos_means'] - dos_mean) ** 2))
doa_std = np.sqrt(
    term1 * np.sum((means['doa_means'] - doa_mean) ** 2))

print(f"DoS Std Dev: {dos_std:.4f}")
print(f"DoA Std Dev: {doa_std:.4f}")

# Calculate required sample size for 95% confidence level and 5% margin of error
z_alpha_score = 1.96  # Z-score for 95% confidence
z_beta_score = 0.842  # Z-score for 80% power

DOS_MDE = (z_alpha_score + z_beta_score) * (dos_std / np.sqrt(len(seeds)))

DOA_MDE = (z_alpha_score + z_beta_score) * (doa_std / np.sqrt(len(seeds)))

print(f"Estimated MDE for DoS: {DOS_MDE:.4f}")
print(f"Estimated MDE for DoA: {DOA_MDE:.4f}")
```
